```python
import pandas as pd
import numpy as np

estoque = pd.read_csv("retail_inventory_ml_apl.csv")
display(estoque.head(5))
```

```python
#informacoes e possiveis tratamentos
print(estoque.duplicated().sum())


display(estoque.info())
display(estoque.describe())
```

```python
#tratando tipo data
estoque['Start Date'] = pd.to_datetime(estoque['Start Date'])
estoque['End Date'] = pd.to_datetime(estoque['End Date'])

estoque.info()
```

```python
#retirando dados

#quantos produtos existem
pdts = estoque['Product No'].nunique()
print(pdts)

#quantos fornecedores
frnc = estoque['Supplier'].nunique()
print(frnc)

#quantas lojas
ljs = estoque['Store'].nunique()
print(ljs)

#quantas categorias
cat = estoque['Product Category'].nunique()
print(cat)

#valor total estoque
vlr_estoq=estoque['Cost of Stocks'].sum()
print(f'{vlr_estoq:,.2f}')

#valor potencial de venda
vlr=estoque['Stocks Selling Amount'].sum()
print(f'{vlr:,.2f}')

#Criando coluna lucro
estoque['Profit'] = estoque['Stocks Selling Amount'] - estoque['Cost of Stocks']
```

```python
#Analises
#categoria mais lucrativa
cat_lucro = estoque[['Product Category', 'Profit']].groupby('Product Category').sum()
cat_lucro = cat_lucro.sort_values(by= 'Profit', ascending= False)
#display(cat_lucro)

#segmentos mais lucartivos
seg_lucro = estoque[['Product Segment', 'Profit']].groupby('Product Segment').sum()
seg_lucro = seg_lucro.sort_values(by='Profit', ascending=False)
#display(seg_lucro)

#lucro subcategoria
subc_lucro = estoque[['Product Subcategory', 'Profit']].groupby('Product Subcategory').sum()
subc_lucro = subc_lucro.sort_values(by='Profit', ascending= False)
#display(subc_lucro)
```

```python
#fonecedores
#fornecedor que gera mais lucro
forn_lucro = estoque[['Supplier', 'Profit']].groupby('Supplier').sum()
forn_lucro = forn_lucro.sort_values(by= 'Supplier', ascending=False)
#display(forn_lucro)

#maior estoque
estoq_forn = estoque[['Supplier', 'Stock Status']].groupby('Supplier').count()
estoq_forn = estoq_forn.sort_values(by='Stock Status', ascending=False)
#display(estoq_forn)

#qual fornecedor possui mais produtos
prod_forn = estoque[['Supplier', 'Product No']].groupby('Supplier').count()
prod_forn = prod_forn.sort_values(by= 'Product No', ascending=False)
#display(prod_forn)

#lojas
#loja com maior estoque
estoq_loja = estoque[['Store', 'Qty on hand']].groupby('Store').sum()
estoq_loja = estoq_loja.sort_values(by='Qty on hand', ascending=False)
#display(estoq_loja)

#loja com maior valor em vendas
loja_venda_estoque = estoque[['Store', 'Stocks Selling Amount']].groupby('Store', as_index= False).sum()
loja_venda_estoque = loja_venda_estoque.sort_values(by='Stocks Selling Amount', ascending=False)
#display(loja_venda_estoque)

#loja mais lucrativa
loja_lucro = estoque[['Store', 'Profit']].groupby('Store').sum()
loja_lucro = loja_lucro.sort_values(by= 'Profit', ascending=False)
display(loja_lucro)
```
