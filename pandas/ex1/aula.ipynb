```python
#tratamento de dados

#excluindo coluna
df_analise = df_vendas.drop(columns= ['Data_Base'])

#tratando os NaN
#df_analise = df_analise.dropna()  => exclui linhas com um ou mais NaN

#troca os valores nulos pela palavra 'online'
df_analise['Loja'] = df_analise['Loja'].fillna('Online')

#mudando o formato da data
df_analise['Data'] = pd.to_datetime(df_analise['Data'], format='%Y-%m-%d')

#padronizacao
df_analise['Loja'] = df_analise['Loja'].str.strip()
df_analise['Loja'] = df_analise['Loja'].str.title()

#remover duplicatas
df_analise = df_analise.drop_duplicates(subset=['ID_Pedido'], keep='last')

display(df_analise)
display(df_analise.info())
```

```python
#criar novas colunas
#faturamente
df_analise['Faturamento'] = df_analise['Qtd'] * df_analise['Preco_Unitario']

#forma de venda
df_analise['Formua de Venda'] = np.where(df_analise['Loja'] == 'Online', 'Online', 'Presencial')
                                        #np.where(condicao, se for vdd, se for falsa)

#regiao
dic_regioes = {
    'São Paulo': 'Sudeste',
    'Belo Horizonte': 'Sudeste',
    'Online': 'Online',
    'Rio De Janeiro': 'Sudeste',
    'Salvador': 'Nordeste',
    'Recife': 'Nordeste',
    'Curitiba': 'Sul',
    'Porto Alegre': 'Sul'
}

df_analise['Regiao'] = df_analise['Loja'].map(dic_regioes)
display(df_analise)
```

```python
#analise =>
#.loc
#                    [indicie, coluna]
loja = df_analise.loc[3, 'Loja']
#print(loja)

#.iloc ==> posicoes
loja = df_analise.iloc[4,4]
#print(loja)

#condicional (linha toda)
df_id_pedido_4 = df_analise[df_analise['ID_Pedido'] == 4]
#print(df_id_pedido_4)

#condicional (so um dado)
df_id_pedido_04 = df_analise[df_analise['ID_Pedido'] == 4].values[0]
#print(df_id_pedido_04)

#exportando
df_vendas_sp= df_analise[df_analise['Loja'] =='São Paulo']
df_vendas_sp.to_csv("Vendas_SP.csv")

#exportando vendas 2024
df_vendas_2024 = df_analise[df_analise['Data'].dt.year == 2024]
#condicional
df_vendas_2024 = df_analise[df_analise['Data'] >= '2024-01-01']

#dupla condicional
condicao =  df_analise['Produto'] == 'Mouse Gamer'
condicao2 =df_analise['Regiao'] == 'Sul'

df_vendas_Mouse = df_analise[condicao & condicao2]
```

```python
#analise por agrupamento
analise_lojas= df_analise[['Loja', 'Faturamento']].groupby('Loja', as_index=False).sum()
#colocando em ordem
analise_lojas = analise_lojas.sort_values(by='Faturamento', ascending= False)
#arrumando indicies
analise_lojas = analise_lojas.sort_index()
#arrumando o visual
analise_lojas['Faturamento'] = analise_lojas['Faturamento'].map('R${:,.2f}'.format)
#display(analise_lojas)

#rank de produto vendidos online
df_vendas_online = df_analise[df_analise['Loja'] == 'Online']

df_analise_vendas_online = df_vendas_online[['Produto','Qtd']].groupby('Produto').sum()

df_analise_vendas_online = df_analise_vendas_online.sort_values(by= 'Qtd', ascending=False)

df_analise_vendas_online= df_analise_vendas_online.rename(columns={'Qtd' : 'Vendas Totais'})
display(df_analise_vendas_online)
```

```python
#quais gerentes bateram a meta 2023
df_meta = df_analise[(df_analise['Data'].dt.year == 2023) & (df_analise['Data'].dt.month == 1)]

df_meta = df_meta[['Loja', 'Faturamento']].groupby('Loja',as_index= False).sum()

df_meta = df_meta.merge(df_gerentes, on = 'Loja', how = 'left')

df_meta['Bateu Meta'] = np.where(df_meta['Faturamento'] >= df_meta['Meta_Mensal'], 'Sim', 'Nao')
display(df_meta)
```
