```python
#Analise de dados chocolate
import pandas as pd
import numpy as np

df_sales = pd.read_csv("Chocolate_Sales.csv")
display(df_sales)
```

```python
print(df_sales.duplicated().sum())

print(df_sales.isnull().sum())

display(df_sales.info())

qtd_prod=df_sales['Product'].nunique()
print(qtd_prod)
```

```python
#price_null=df_sales[df_sales['Price_per_Box'].isna()]
#display(price_null)

df_sales['Amount'] = ( df_sales['Amount'] .str.replace('$', '', regex=False) .str.replace(',', '', regex=False) .astype(float) )
df_sales['Amount'] = df_sales['Amount'].astype(float)

df_sales= df_sales.dropna()

display(df_sales.info())
```

```python
#Analise descritiva vendas internacionais de chocolate
#fato pais
maior_receita_pais = df_sales[["Country", "Amount"]].groupby('Country').sum()
maior_receita_pais = maior_receita_pais.sort_values(by="Country", ascending=True)
maior_receita_pais['Amount'] =maior_receita_pais['Amount'].map("R${:,.2f}".format)

display(maior_receita_pais)
```

```python
#produto com mais lucro em cada pais
produto_pais=df_sales[["Country", "Product", "Amount"]].groupby(['Country', 'Product']).sum()
produto_pais = produto_pais.sort_values(by=['Country','Amount'], ascending=False)
produto_pais['Amount'] =produto_pais['Amount'].map("R${:,.2f}".format)
#display(produto_pais)

#produto mais vendido (caixas enviadas) por pais
caixas_env= df_sales[["Country", "Product", "Boxes_Shipped","Amount"]].groupby(['Country', 'Product']).sum()
caixas_env= caixas_env.sort_values(by=['Country','Boxes_Shipped'], ascending=False)
caixas_env['Amount'] =caixas_env['Amount'].map("R${:,.2f}".format)
display(caixas_env)
```

```python
#Fato Marketing
#Qual pais gastou mais com marketing

marketing = df_sales[['Marketing_Spend', 'Country']].groupby('Country').sum()
marketing = marketing.sort_values(by='Country', ascending=True)
marketing['Marketing_Spend'] =marketing['Marketing_Spend'].map("R${:,.2f}".format)
display(marketing)

#lucro total paises
lucro_pais = df_sales[['Amount', 'Country']].groupby('Country').sum()
lucro_pais = lucro_pais.sort_values(by='Country', ascending=True)
lucro_pais['Amount'] =lucro_pais['Amount'].map("R${:,.2f}".format)
display(lucro_pais)
```

```python
#fato desconto
#faixas de descontos
#med_desc = df_sales['Discount_Pct'].mean()
#print(med_desc)


faixas = [0, 8.3, 13.1, 17.5, 38.3]

cat = ["Baixo", "Medio", "Alto", 'Muito alto']

df_sales["Desconto"] = pd.cut(df_sales["Discount_Pct"], bins= faixas, labels=cat)

display(df_sales.head())
```

```python
#A faixa de desconto influencia nos ganhos
desconto_ganhos = df_sales[['Desconto', 'Amount']].groupby('Desconto').sum()
desconto_ganhos = desconto_ganhos.sort_values(by='Amount', ascending= False)
desconto_ganhos['Amount'] =desconto_ganhos['Amount'].map("R${:,.2f}".format)
display(desconto_ganhos)
```

```python
desconto_canal = df_sales[['Desconto', 'Channel', 'Amount']].groupby(['Desconto', 'Channel']).sum()
desconto_canal = desconto_canal.sort_values(by='Amount', ascending= False)
desconto_canal['Amount'] =desconto_canal['Amount'].map("R${:,.2f}".format)
display(desconto_canal)
```
